# `equal_width=True` on mosaic plots

`type="mosaic"` normally makes each column's width proportional to that
`x` value's marginal frequency. `equal_width=True` keeps everything else
about mosaic (stacking, color, legend, annotation) but makes every column
the same width instead -- a 100%-stacked bar chart per `x` value.

```python
from symbulate import *
import numpy as np
```

`x` here is `Bernoulli(p=0.9)` -- a heavily skewed marginal (~90%/10%
split) -- so the default mosaic's two columns are visibly different
widths:

```python
np.random.seed(7)
X, Y = RV(Bernoulli(p=0.9) * Bernoulli(p=0.5))
(X & Y).sim(500).plot(type="mosaic", marginal_column=False, suggest=False)
```

Same data, `equal_width=True`: both columns are now the same width --
only the layout changed. Segment heights (each column's own conditional
split of `y`) are identical to the plot above. The title also switches to
"Stacked Plot" (instead of "Mosaic Plot"), so it's visually obvious which
mode is showing even without comparing widths side by side.

```python
np.random.seed(7)
X, Y = RV(Bernoulli(p=0.9) * Bernoulli(p=0.5))
(X & Y).sim(500).plot(
    type="mosaic", equal_width=True, marginal_column=False, suggest=False
)
```

`equal_width` composes with every other mosaic option, unchanged --
here with the default `marginal_column=True` and a heavily skewed
`x = Binomial(4, 0.15)` (5 columns of very unequal size):

```python
np.random.seed(7)
X, Y = RV(Binomial(n=4, p=0.15) * Bernoulli(p=0.4))
sims = (X & Y).sim(600)
sims.plot(type="mosaic", equal_width=True, suggest=False)
```
