# Controlling the plotting window with `xlim`

`Distribution.plot()` accepts three kinds of `xlim`, and they behave the same for `cdf=True` as for the pdf/pmf (`xlim` only picks the x-values; `cdf` decides which curve is drawn):

- **`None`** (default) — full support if the distribution is bounded, otherwise the window holding ~99.8% of the probability.
- **`"zoom"`** — the tightest high-density window, applied even to a bounded distribution.
- **`(low, high)`** — an exact range you choose.

```python
from symbulate import *
```

## Skewed distribution: the default already trims the long tail

```python
LogNormal(0, 1).plot(cdf=True)
```

## `xlim="zoom"` reframes a bounded distribution on its high-probability region

```python
Beta(2, 5).plot(cdf=True, xlim="zoom")
```

## `xlim="zoom"` on a discrete bounded distribution

```python
Binomial(100, 0.5).plot(cdf=True, xlim="zoom")
```

## An exact range with `xlim=(low, high)`

```python
Normal(0, 1).plot(cdf=True, xlim=(-2, 2))
```

## An exact range on a discrete CDF

```python
Binomial(100, 0.5).plot(cdf=True, xlim=(40, 60))
```
