# Plotting a distribution's x-axis limits

`Distribution.plot(xlim=...)` sets the x-axis: leave it out for the natural window, use `"zoom"` to frame the high-probability region, or pass a `(low, high)` tuple for exact limits.

```python
from symbulate import *
```

## Default window

```python
Binomial(100, 0.5).plot()
```

## Zoom to the high-probability region

```python
Binomial(100, 0.5).plot(xlim="zoom")
```

## Exact limits with a tuple

```python
Binomial(100, 0.5).plot(xlim=(30, 70))
```

## Comparing a theoretical curve to simulated data

By default the curve keeps its full support, so the shared axis stretches across the whole range.

```python
RV(Binomial(100, 0.5)).sim(10000).plot()
Binomial(100, 0.5).plot()
```

## The same overlay stays tight with `xlim="zoom"`

```python
RV(Binomial(100, 0.5)).sim(10000).plot()
Binomial(100, 0.5).plot(xlim="zoom")
```

## Unbounded distributions already use a high-probability window

```python
Poisson(20).plot()
```

## `xlim="zoom"` works on any distribution

```python
Gamma(2).plot(xlim="zoom")
```
