# CDF plotting for distributions

`cdf=True` in `Distribution.plot()` plots the cumulative distribution function. The default (`cdf=False`) plots the pdf/pmf.

```python
from symbulate import *
```

## Discrete

```python
Poisson(5).plot(cdf=True)
```

## Continuous

```python
Normal(0, 1).plot(cdf=True)
```

## Default is still the pdf/pmf

```python
Normal(0, 1).plot()
```

```python
Poisson(5).plot()
```

## Works with xlim="zoom" to zoom in

```python
Binomial(100, 0.5).plot(cdf=True, xlim="zoom")
```

## Compare theoretical CDF to the simulated ECDF

```python
RV(Poisson(5)).sim(1000).plot(type="ecdf")
Poisson(5).plot(cdf=True, xlim="zoom")
```
