# CDF plotting for distributions

`type="cdf"` in `Distribution.plot()` plots the cumulative distribution function. Default is still `type="pdf"` for pdf/pmf.

```python
from symbulate import *
```

## Discrete

```python
Poisson(5).plot(type="cdf")
```

## Continuous

```python
Normal(0, 1).plot(type="cdf")
```

## Default is still the pdf/pmf

```python
Normal(0, 1).plot()
```

```python
Poisson(5).plot()
```

## Works with prob= to zoom in

```python
Binomial(100, 0.5).plot(type="cdf", prob=0.5)
```

## Compare theoretical CDF to the simulated ECDF

```python
RV(Poisson(5)).sim(1000).plot(type="ecdf")
Poisson(5).plot(type="cdf", prob=True)
```
