# CDF plotting for distributions

`cdf=True` in `Distribution.plot()` plots the cumulative distribution function `P(X ≤ x)`. The default (`cdf=False`) plots the pdf/pmf.

```python
from symbulate import *
```

## Discrete CDF is a right-continuous step function

```python
Poisson(5).plot(cdf=True)
```

## Continuous CDF is a smooth S-curve

```python
Normal(0, 1).plot(cdf=True)
```

## Default is still the pdf/pmf

```python
Normal(0, 1).plot()
```

## Compare a theoretical CDF to the simulated ECDF

```python
RV(Poisson(5)).sim(1000).plot(type="ecdf")
Poisson(5).plot(cdf=True)
```
