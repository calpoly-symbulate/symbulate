# De Moivre Distribution

`DeMoivre(omega)` is the simplest survival law where age at death is uniform between 0 and a limiting age `omega`.

```python
from symbulate import *
```

## Define a De Moivre distribution and draw an age at death

```python
X = DeMoivre(omega=100)
X.draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `omega / 2`

```python
X.mean(), X.var(), X.sd()
```

## Assigning and simulating a random variable

```python
D = RV(X)
D.sim(10000)
```

## Plot the pdf

Flat: every age up to `omega` is equally likely.

```python
X.plot()
```

## Changing the limiting age omega

A larger `omega` spreads the ages over more years, so the flat density gets wider and lower.

```python
for w in [60, 80, 100, 120]:
    DeMoivre(omega=w).plot()
```

## Plot the CDF

A straight line from 0 to 1

```python
X.plot(cdf=True)
```

## Special Case: It is a reparameterized Uniform

`DeMoivre(omega)` is exactly `Uniform(0, omega)`.

```python
DeMoivre(100).plot()
Uniform(0, 100).plot()
```
