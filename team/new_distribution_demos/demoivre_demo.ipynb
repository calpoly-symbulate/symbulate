# De Moivre Distribution

`DeMoivre(omega)` is the simplest survival law where age at death is uniform between 0 and a limiting age `omega`.

```python
from symbulate import *
```

## Draw an age at death from a De Moivre distribution

```python
DeMoivre(omega=100).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `omega / 2`.

```python
DeMoivre(omega=100).mean(), DeMoivre(omega=100).var(), DeMoivre(omega=100).sd()
```

## Assign and simulate a random variable

```python
D = RV(DeMoivre(omega=100))
D.sim(10000)
```

## Plot the pdf

Flat: every age up to `omega` is equally likely.

```python
DeMoivre(omega=100).plot()
```

## Changing the limiting age omega

A larger `omega` spreads the ages over more years, so the flat density gets wider and lower.

```python
for w in [60, 80, 100, 120]:
    DeMoivre(omega=w).plot()
```

## Plot the CDF

A straight line from 0 to 1.

```python
DeMoivre(omega=100).plot(cdf=True)
```

## Relationship: a reparameterized Uniform

`DeMoivre(omega)` is exactly `Uniform(0, omega)` -- age at death is uniform on `[0, omega]`.

```python
DeMoivre(omega=100).plot()
Uniform(a=0, b=100).plot()
```
