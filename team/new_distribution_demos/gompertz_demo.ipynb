# Gompertz Distribution

`Gompertz(shape, scale)` is a continuous distribution where hazard rate rises exponentially with age.

```python
from symbulate import *
```

## Define a Gompertz distribution and draw a value

```python
X = Gompertz(shape=1.5, scale=2)
X.draw()
```

## Theoretical mean, variance, and standard deviation

```python
X.mean(), X.var(), X.sd()
```

## Assigning and simulating a random variable

```python
G = RV(X)
G.sim(10000)
```

## Plot the pdf

```python
X.plot()
```

## Plot the CDF

```python
X.plot(cdf=True)
```

## Shape parameter changes the curve

A larger shape pushes mass toward smaller values -- mortality sets in sooner.

```python
for k in [0.5, 1, 2, 4]:
    Gompertz(shape=k, scale=1).plot()
```
