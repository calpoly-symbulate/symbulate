# Gompertz Distribution

`Gompertz(shape, scale)` is a continuous distribution where the hazard rate rises exponentially with age.

```python
from symbulate import *
```

## Draw a value from a Gompertz distribution

```python
Gompertz(shape=1.5, scale=2).draw()
```

## Theoretical mean, variance, and standard deviation

```python
Gompertz(shape=1.5, scale=2).mean(), Gompertz(shape=1.5, scale=2).var(), Gompertz(shape=1.5, scale=2).sd()
```

## Assign and simulate a random variable

```python
G = RV(Gompertz(shape=1.5, scale=2))
G.sim(10000)
```

## Plot the pdf

```python
Gompertz(shape=1.5, scale=2).plot()
```

## Plot the CDF

```python
Gompertz(shape=1.5, scale=2).plot(cdf=True)
```

## Shape parameter changes the curve

A larger shape pushes mass toward smaller values -- mortality sets in sooner.

```python
for k in [0.5, 1, 2, 4]:
    Gompertz(shape=k, scale=1).plot()
```

## Relationship: Gompertz is a Makeham with no accident term

The `Makeham` distribution is a Gompertz plus a constant, age-independent hazard term. Setting that term to zero (`makeham=0`) recovers the Gompertz exactly.

```python
Gompertz(shape=1.5, scale=2).plot()
Makeham(shape=1.5, makeham=0, scale=2).plot()
```
