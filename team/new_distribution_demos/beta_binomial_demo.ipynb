# Beta-Binomial Distribution

`BetaBinomial(n, a, b)` counts the number of successes in `n` trials when the success probability is not fixed but is itself random.

```python
from symbulate import *
```

## Draw a value from a beta-binomial distribution

```python
BetaBinomial(n=10, a=2, b=3).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `n * a / (a + b)`.

```python
BetaBinomial(n=10, a=2, b=3).mean(), BetaBinomial(n=10, a=2, b=3).var(), BetaBinomial(n=10, a=2, b=3).sd()
```

## Assign and simulate a random variable

```python
B = RV(BetaBinomial(n=10, a=2, b=3))
B.sim(10000)
```

## Plot the pmf

```python
BetaBinomial(n=10, a=2, b=3).plot()
```

## Plot the CDF

```python
BetaBinomial(n=10, a=2, b=3).plot(cdf=True)
```

## Relationship: `a = b = 1` is a discrete uniform

```python
BetaBinomial(n=10, a=1, b=1).plot()
DiscreteUniform(a=0, b=10).plot()
```

## Relationship: large `a` and `b` approach a Binomial

```python
BetaBinomial(n=10, a=300, b=700).plot()
Binomial(n=10, p=0.3).plot()
```
