# Exponentially Modified Gaussian Distribution

`ExponentiallyModifiedGaussian(mean, sd, rate)` is the distribution of a normal value **plus** an independent exponential value.

```python
from symbulate import *
```

## Draw a value from an exponentially modified Gaussian distribution

```python
ExponentiallyModifiedGaussian(mean = 5, sd= 2, rate = .5).draw()
```

## Theoretical mean, variance, and standard deviation

The two pieces are independent, so means add and variances add: `mean + 1 / rate` and `sd ** 2 + 1 / rate ** 2`.

TODO: show that `X.mean()` is *not* the `mean` argument -- the exponential part shifts the center to the right.

```python
X = ExponentiallyModifiedGaussian(mean = 5, sd= 2, rate = .5)
X.mean(), X.var(), X.sd()
```

## Assign and simulate a random variable

```python
X = RV(ExponentiallyModifiedGaussian(mean=0, sd=1, rate=1))
X.sim(10000)
```

## Plot the pdf

```python
ExponentiallyModifiedGaussian(mean = 5, sd= 2, rate = .5).plot()
```

## Plot the CDF

```python
ExponentiallyModifiedGaussian(mean = 5, sd= 2, rate = .5).plot(cdf=True)
```
