# Skew-t Distribution

`SkewT(shape1, shape2, loc, scale)` is a heavy-tailed, asymmetric bell curve -- Student's t-distribution with one tail allowed to be longer than the other.
When they are equal the curve is symmetric and equals a Student's t with `shape1 + shape2` degrees of freedom. When `shape1 > shape2` the longer tail is on the right; when `shape1 < shape2` it is on the left.

```python
from symbulate import *
```

## Draw a value

```python
SkewT(shape1=5, shape2=2).draw()
```

## Theoretical mean, variance, and standard deviation

```python
SkewT(shape1=5, shape2=2).mean(), SkewT(shape1=5, shape2=2).var(), SkewT(shape1=5, shape2=2).sd()
```

## Assign and simulate a random variable

```python
T = RV(SkewT(shape1=5, shape2=2))
T.sim(10000)
```

## Plot the pdf

The peak is pulled to the left of center and the right tail stretches out -- a right-skewed, heavy-tailed curve.

```python
SkewT(shape1=5, shape2=2).plot()
```

## Plot the CDF

```python
SkewT(shape1=5, shape2=2).plot(cdf=True)
```

## The two shapes set the direction of the skew

Holding the total `shape1 + shape2` fixed, moving weight from one shape to the other tips the skew from left to right. The middle case (`shape1 == shape2`) is symmetric.

```python
for shape1, shape2 in [(2, 6), (3, 5), (4, 4), (5, 3), (6, 2)]:
    SkewT(shape1=shape1, shape2=shape2).plot()
```

## Relationship: equal shapes give a Student's t

When `shape1 == shape2`, the skew-t is symmetric and is exactly a `StudentT` with `shape1 + shape2` degrees of freedom. Here `SkewT(4, 4)` matches `StudentT(df=8)`.

```python
SkewT(shape1=4, shape2=4).plot()
StudentT(df=8).plot()
```

## Limit: large equal shapes approach the normal

As both shapes grow the tails lighten, and a large symmetric skew-t is nearly a standard normal -- the same way a high-`df` Student's t is.

```python
SkewT(shape1=20, shape2=20).plot()
Normal(mean=0, sd=1).plot()
```
