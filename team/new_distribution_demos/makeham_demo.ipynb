# Makeham (Gompertz-Makeham) Distribution

`Makeham(shape, makeham, scale)` is a continuous distribution used in actuarial science to model mortality. It is the Gompertz law plus a constant, age-independent hazard term (the "accident" term). Setting `makeham=0` recovers the plain Gompertz distribution.

```python
from symbulate import *
```

## Define a Makeham distribution and draw a value

```python
X = Makeham(shape=1.5, makeham=0.3, scale=2)
X.draw()
```

## Theoretical mean, variance, and standard deviation

```python
X.mean(), X.var(), X.sd()
```

## Assigning and simulating a random variable

```python
M = RV(X)
M.sim(10000)
```

## Plot the pdf

```python
X.plot()
```

## Plot the CDF

```python
X.plot(cdf=True)
```

## The makeham term adds a constant floor to mortality

With `makeham=0` this is exactly a Gompertz. Larger values raise the age-independent risk, shifting mass toward smaller values.

```python
for a in [0, 0.3, 1, 2]:
    Makeham(shape=1.5, makeham=a, scale=1).plot()
```
