# Burr (Burr Type XII) Distribution

`Burr(a, b, scale)` is a heavy-tailed continuous distribution on `[0, inf)`. Often used in actuarial science and economics.

```python
from symbulate import *
```

## Draw a value from a Burr distribution

```python
Burr(a=3, b=2, scale=1).draw()
```

## Theoretical mean, variance, and standard deviation

The `k`-th moment is finite only when `a * b > k`, so a heavy-tailed Burr can have an infinite mean or variance.

```python
Burr(a=3, b=2, scale=1).mean(), Burr(a=3, b=2, scale=1).var(), Burr(a=3, b=2, scale=1).sd()
```

## Assign and simulate a random variable

```python
B = RV(Burr(a=3, b=2, scale=1))
B.sim(10000)
```

## Plot the pdf

```python
Burr(a=3, b=2, scale=1).plot()
```

## Plot the CDF

```python
Burr(a=3, b=2, scale=1).plot(cdf=True)
```

## Shape parameters change the curve

`a` shapes the body near the origin; `b` controls the tail.

```python
for a, b in [(1, 2), (2, 1), (3, 2), (4, 4)]:
    Burr(a=a, b=b, scale=1).plot()
```

## Relationship: `a = 1` is a Pareto

With `a = 1` the Burr is a Pareto (Type II / Lomax) on `[0, inf)`. Shifting those values up by the scale gives a standard Pareto (Type I). The Burr generalizes the Pareto.

```python
P = RV(Burr(a=1, b=2, scale=1))
(P + 1).sim(10000).plot(bins=100)
Pareto(b=2, scale=1).plot()
```
