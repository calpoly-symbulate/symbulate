# Weibull Distribution

`Weibull(shape, scale)` is a continuous distribution for time-to-failure.

```python
from symbulate import *
```

## Define a Weibull distribution and draw a value

```python
X = Weibull(shape=3, scale=2)
X.draw()
```

## Theoretical mean, variance, and standard deviation

```python
X.mean(), X.var(), X.sd()
```

## Assigning and simulating a random variable

```python
W1 = RV(X)
W1.sim(10000)
```

## Plot the pdf

```python
X.plot()
```

## Special case: shape = 1 is the Exponential distribution

```python
Weibull(shape=1, scale=2).plot()
Exponential(rate=0.5).plot()
```

## Special case: shape = 2 is the Rayleigh distribution

```python
Weibull(shape=2, scale=2**0.5).plot()
Rayleigh().plot()
```

## Shape parameter changes the curve

```python
for k in [0.5, 1, 1.5, 3]:
    Weibull(shape=k, scale=1).plot()
```
