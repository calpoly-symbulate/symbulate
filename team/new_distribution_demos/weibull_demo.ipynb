# Weibull Distribution

`Weibull(shape, scale)` is a continuous distribution for time-to-failure.

```python
from symbulate import *
```

## Draw a value from a Weibull distribution

```python
Weibull(shape=3, scale=2).draw()
```

## Theoretical mean, variance, and standard deviation

```python
Weibull(shape=3, scale=2).mean(), Weibull(shape=3, scale=2).var(), Weibull(shape=3, scale=2).sd()
```

## Assign and simulate a random variable

```python
W1 = RV(Weibull(shape=3, scale=2))
W1.sim(10000)
```

## Plot the pdf

```python
Weibull(shape=3, scale=2).plot()
```

## Plot the CDF

```python
Weibull(shape=3, scale=2).plot(cdf=True)
```

## Special case: shape = 1 is the Exponential distribution

With `shape = 1` the failure rate is constant, so the Weibull reduces to an `Exponential` with the same scale (`rate = 1 / scale`).

```python
Weibull(shape=1, scale=2).plot()
Exponential(scale=2).plot()
```

## Special case: shape = 2 is the Rayleigh distribution

With `shape = 2` and `scale = sqrt(2)`, the Weibull is exactly a standard `Rayleigh`.

```python
Weibull(shape=2, scale=2 ** 0.5).plot()
Rayleigh().plot()
```

## Shape parameter changes the curve

```python
for k in [0.5, 1, 1.5, 3]:
    Weibull(shape=k, scale=1).plot()
```
