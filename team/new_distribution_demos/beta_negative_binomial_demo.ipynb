# Beta-Negative Binomial (Beta-Pascal) Distribution

`BetaNegativeBinomial(r, shape1, shape2)` counts the number of failures before the `r`-th success with a probability of success that is randomly drawn from a `Beta` distribution with shape parameters `shape1` and `shape2`.

```python
from symbulate import *
```

## Draw a value from a beta-negative binomial distribution

```python
BetaNegativeBinomial(r=5, shape1=3, shape2=2).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `r * shape2 / (shape1 - 1)` and is finite only when `shape1 > 1`.

```python
BetaNegativeBinomial(r=5, shape1=3, shape2=2).mean(), BetaNegativeBinomial(r=5, shape1=3, shape2=2).var(), BetaNegativeBinomial(r=5, shape1=3, shape2=2).sd()
```

## Assign and simulate a random variable

```python
B = RV(BetaNegativeBinomial(r=5, shape1=3, shape2=2))
B.sim(10000)
```

## Plot the pmf

```python
BetaNegativeBinomial(r=5, shape1=4, shape2=3).plot()
```

## Plot the CDF

```python
BetaNegativeBinomial(r=5, shape1=4, shape2=3).plot(cdf=True)
```

## Relationship: large `shape1` and `shape2` approach a Pascal

As `shape1` and `shape2` grow with the ratio `shape1 / (shape1 + shape2)` fixed, the beta concentrates on a single probability `p` and the beta-negative binomial approaches `Pascal(r, p)`.

```python
BetaNegativeBinomial(r=5, shape1=400, shape2=300).plot()
Pascal(r=5, p=4/7).plot()
```

## Special case: `r = 1` is the beta-geometric

With `r = 1` the distribution counts the number of failures before the *first* success when the success probability is random (the beta-geometric distribution).

```python
BetaNegativeBinomial(r=1, shape1=3, shape2=2).plot()
```
