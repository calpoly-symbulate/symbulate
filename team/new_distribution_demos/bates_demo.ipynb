# Bates Distribution

`Bates(n, a, b)` is the distribution of the average of `n` independent `Uniform(a, b)` random variables.

```python
from symbulate import *
```

## Draw a value from a Bates distribution

```python
Bates(n=5).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is the uniform mean `(a + b) / 2`; the variance is `(b - a)**2 / (12 * n)`, the single-uniform variance shrunk by a factor of `n`.

```python
Bates(n=5).mean(), Bates(n=5).var(), Bates(n=5).sd()
```

## Assign and simulate a random variable

```python
B = RV(Bates(n=5))
B.sim(10000)
```

## Plot the pdf

```python
Bates(n=5).plot()
```

## Plot the CDF

```python
Bates(n=5).plot(cdf=True)
```

## The central limit theorem

As `n` grows, the average of `n` uniforms goes from flat (`n = 1`) to triangular (`n = 2`) to a tight bell.

```python
for n in [1, 2, 5, 20]:
    Bates(n=n).plot()
```

## Special case: `n = 1` is the Uniform

The average of a single uniform is just that uniform, so the two densities coincide.

```python
Bates(n=1, a=0, b=1).plot()
Uniform(a=0, b=1).plot()
```

## Approaches a Normal (the CLT)

For large `n`, `Bates(n)` is very close to a `Normal` with the same mean and variance.

```python
Bates(n=20).plot()
Normal(mean=0.5, sd=(1 / (12 * 20)) ** 0.5).plot()
```
