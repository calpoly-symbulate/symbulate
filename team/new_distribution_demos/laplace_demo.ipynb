# Laplace (Double Exponential) Distribution

`Laplace(loc, scale)` is a continuous distribution that looks like two exponential distributions placed back to back: a sharp peak at the center and heavier tails than the normal.

```python
from symbulate import *
```

## Draw a value from a Laplace distribution

```python
Laplace(loc=0, scale=1).draw()
```

## Theoretical mean, variance, and standard deviation

The mean equals `loc`; the standard deviation is `scale * sqrt(2)`.

```python
Laplace(loc=0, scale=1).mean(), Laplace(loc=0, scale=1).var(), Laplace(loc=0, scale=1).sd()
```

## Assign and simulate a random variable

```python
L = RV(Laplace(loc=0, scale=1))
L.sim(10000)
```

## Plot the pdf

Symmetric about `loc`, with a sharp peak at the center.

```python
Laplace(loc=0, scale=1).plot()
```

## Plot the CDF

```python
Laplace(loc=0, scale=1).plot(cdf=True)
```

## Scale parameter changes the spread

```python
for s in [0.5, 1, 2, 4]:
    Laplace(loc=0, scale=s).plot()
```

## Relationship: the difference of two Exponentials

If `E1` and `E2` are independent `Exponential(rate=1)` variables, then `E1 - E2` is `Laplace(0, 1)` -- the two back-to-back exponentials the shape suggests.

```python
E1, E2 = RV(Exponential(rate=1) * Exponential(rate=1))
(E1 - E2).sim(10000).plot()
Laplace(loc=0, scale=1).plot()
```
