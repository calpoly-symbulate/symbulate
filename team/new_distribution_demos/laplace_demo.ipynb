# Laplace (Double Exponential) Distribution

`Laplace(loc, scale)` is a continuous distribution that looks like two exponential distributions placed back to back: a sharp peak at the center and heavier tails than the normal.

```python
from symbulate import *
```

## Define a Laplace distribution and draw a value

```python
X = Laplace(loc=0, scale=1)
X.draw()
```

## Theoretical mean, variance, and standard deviation

The mean equals `loc`; the standard deviation is `scale * sqrt(2)`.

```python
X.mean(), X.var(), X.sd()
```

## Assigning and simulating a random variable

```python
L = RV(X)
L.sim(10000)
```

## Plot the pdf

Symmetric about `loc`, with a sharp peak at the center.

```python
X.plot()
```

## Plot the CDF

```python
X.plot(cdf=True)
```

## Scale parameter changes the spread

```python
for s in [0.5, 1, 2, 4]:
    Laplace(loc=0, scale=s).plot()
```
