# Lomax (Pareto Type II) Distribution

`Lomax(b, scale)` is a heavy-tailed continuous distribution on `[0, inf)`. It is the Pareto distribution shifted to start at 0 instead of at its scale.

```python
from symbulate import *
```

## Draw a value from a Lomax distribution

```python
Lomax(b=3, scale=1).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is finite only when `b > 1`, and the variance only when `b > 2`.

```python
Lomax(b=3, scale=1).mean(), Lomax(b=3, scale=1).var(), Lomax(b=3, scale=1).sd()
```

## Assign and simulate a random variable

```python
L = RV(Lomax(b=3, scale=1))
L.sim(10000)
```

## Plot the pdf

The density peaks at 0 and decays with a heavy right tail.

```python
Lomax(b=3, scale=1).plot()
```

## Plot the CDF

```python
Lomax(b=3, scale=1).plot(cdf=True)
```

## Shape parameter changes the tail

```python
for b in [3, 5, 8, 12]:
    Lomax(b=b, scale=1).plot()
```

## Relationship: Lomax is the `a = 1` Burr

A `Lomax(b, scale)` is exactly a `Burr(a=1, b, scale)`.

```python
Burr(a=1, b=3, scale=1).plot()
Lomax(b=3, scale=1).plot()
```

## Relationship: shifting by the scale gives a Pareto

Adding the scale to a `Lomax(b, scale)` gives a `Pareto(b, scale)` (Type I).

```python
L = RV(Lomax(b=3, scale=1))
(L + 1).sim(10000).plot(bins=100)
Pareto(b=3, scale=1).plot()
```
