# Generalized Extreme Value (GEV) Distribution

`GEV(loc, scale, shape)` is a continuous distribution for the maximum of many observations.

```python
from symbulate import *
```

## Define a GEV distribution and draw a value

```python
X = GEV(loc=0, scale=1, shape=0.2)
X.draw()
```

## Theoretical mean, variance, and standard deviation

```python
X.mean(), X.var(), X.sd()
```

## Assigning and simulating a random variable

```python
G = RV(X)
G.sim(10000)
```

## Plot the pdf

```python
X.plot()
```

## Plot the CDF

```python
X.plot(cdf=True)
```

## Special Cases: shape parameter has three laws

`shape < 0` is reverse-Weibull (bounded above)

`shape = 0` is Gumbel (unbounded)

`shape > 0` is Frechet (bounded below --> right-tail)

```python
for xi in [-0.3, 0, 0.3]:
    GEV(loc=0, scale=1, shape=xi).plot()
```
