# Generalized Extreme Value (GEV) Distribution

`GEV(loc, scale, shape)` is a continuous distribution for the maximum of many observations.

```python
from symbulate import *
```

## Draw a value from a GEV distribution

```python
GEV(loc=0, scale=1, shape=0.2).draw()
```

## Theoretical mean, variance, and standard deviation

```python
GEV(loc=0, scale=1, shape=0.2).mean(), GEV(loc=0, scale=1, shape=0.2).var(), GEV(loc=0, scale=1, shape=0.2).sd()
```

## Assign and simulate a random variable

```python
G = RV(GEV(loc=0, scale=1, shape=0.2))
G.sim(10000)
```

## Plot the pdf

```python
GEV(loc=0, scale=1, shape=0.2).plot()
```

## Plot the CDF

```python
GEV(loc=0, scale=1, shape=0.2).plot(cdf=True)
```

## Relationship: the shape parameter selects the three extreme-value laws

A single `shape` (the extreme-value index) unifies the three classical laws:

- `shape < 0` is the reverse-Weibull distribution (bounded above)
- `shape = 0` is the Gumbel distribution (unbounded)
- `shape > 0` is the Frechet distribution (bounded below, heavy right tail)

```python
for xi in [-0.3, 0, 0.3]:
    GEV(loc=0, scale=1, shape=xi).plot()
```
