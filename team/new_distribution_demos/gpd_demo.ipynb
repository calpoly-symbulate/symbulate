# Generalized Pareto Distribution

```python
from symbulate import *
```

## Draw a value from a GPD distribution

```python
GPD(loc=0, scale=1, shape=0.2).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is finite only when `shape < 1`, and the variance only when `shape < 1/2`.

```python
GPD(loc=0, scale=1, shape=0.2).mean(), GPD(loc=0, scale=1, shape=0.2).var(), GPD(loc=0, scale=1, shape=0.2).sd()
```

## Assign and simulate a random variable

```python
X = RV(GPD(loc=0, scale=1, shape=0.2))
X.sim(10000)
```

## Plot the pdf

The density peaks at `loc` and decays to the right.

```python
GPD(loc=0, scale=1, shape=0.2).plot()
```

## Plot the CDF

```python
GPD(loc=0, scale=1, shape=0.2).plot(cdf=True)
```

## Relationship: `shape = 0` is an exponential


```python
E = RV(Exponential(scale=2))
E.sim(10000).plot()
GPD(loc=0, scale=2, shape=0).plot()
```

## Relationship: a positive `shape` is a Pareto

A `Pareto(shape, scale=s)` is exactly a `GPD(loc=s, scale=s/shape, shape=1/shape)`. We simulate Pareto values and overlay the GPD density.

```python
P = RV(Pareto(shape=5, scale=1))
P.sim(10000).plot(bins=60)
GPD(loc=1, scale=0.2, shape=0.2).plot()
```
