# Random Walks

Flip a coin. Heads, take a step forward; tails, take a step back. Keep going. Where do you
end up?

That is a **random walk** -- the running total of independent random steps:

$$X[n] = X[n-1] + \text{step}[n]$$

It is usually the first random process anyone meets, because two big ideas live inside it.
One is **gambler's ruin**: if you bet £1 at a time, will you hit your target before you go
broke? The other is the **central limit theorem**: $X[n]$ is a sum of independent pieces, so
its spread grows in a very particular way.

```python
from symbulate import *
```

## Create a walk and take a look at one path

`p=0.5` gives the fair coin-flip walk: each step is $+1$ or $-1$, equally likely.

```python
X = RandomWalk(p=0.5)
X.draw().plot()
```

Every path starts at 0 and moves by exactly 1 each step. Run the cell a few times -- some
paths wander up, some down, some hover near zero. None of them is "typical".

## Several paths at once

```python
X.sim(20).plot()
```

The fan shape is the thing to notice. Paths spread out as time goes on, but not as fast as
you might guess -- see below.

## Where does it end up after n steps?

`X[n]` is the random variable "position after $n$ steps", so it can be simulated like
anything else.

```python
X[100].sim(10000).plot()
```

A bell curve, centred at 0. That is the central limit theorem showing up: the position is a
sum of 100 independent steps, so it looks normal even though a single step is just $\pm 1$.

## How fast does it spread?

Here is the fact worth remembering. For the fair walk, the **variance after $n$ steps is
exactly $n$**.

```python
for n in [10, 50, 100]:
    print("after", n, "steps:  variance", round(X[n].sim(10000).var(), 1))
```

Variance $n$ means a standard deviation of $\sqrt{n}$. So after 100 steps you are typically
about 10 away from where you started -- not 100. Quadrupling the number of steps only
doubles how far you drift. That is why the fan above widens so slowly.

## A biased coin

`p` is the chance of stepping up. Anything other than 0.5 gives the walk a drift, and the
average position after $n$ steps is $n(2p - 1)$.

```python
RandomWalk(p=0.6).sim(20).plot()
```

```python
Y = RandomWalk(p=0.6)
print("average position after 50 steps:", round(Y[50].sim(10000).mean(), 2))
print("theory: 50 * (2*0.6 - 1) =", round(50 * (2 * 0.6 - 1), 2))
```

## Gambler's ruin

Now the classic problem. You walk into a casino with £10 and decide to leave when you either
reach £20 or run out of money. You bet £1 on a fair coin each time. What are your chances?

Start the walk at 10 with `initial`, then follow the path until it hits 0 or 20.

```python
def reaches_goal(path):
    n = 0
    while 0 < path[n] < 20:
        n = n + 1
    return path[n] == 20


G = RandomWalk(p=0.5, initial=10)
G.apply(reaches_goal).sim(2000).mean()
```

About half the time -- which makes sense, since 10 is halfway between 0 and 20.

The general answer for a fair walk is that your chance of reaching the target is just your
starting money divided by the target. Starting with £5 instead should give about 0.25:

```python
RandomWalk(p=0.5, initial=5).apply(reaches_goal).sim(2000).mean()
```

## How long does it take?

The same idea, but counting steps instead of checking where it ended. The theoretical answer
starting from £10 is $10 \times (20 - 10) = 100$ bets.

```python
def steps_until_done(path):
    n = 0
    while 0 < path[n] < 20:
        n = n + 1
    return n


G.apply(steps_until_done).sim(2000).mean()
```

A hundred bets on average -- even though the game is a coin flip and you are only £10 from
either end. Random walks take a surprisingly long time to get anywhere.

## Steps that are not just plus or minus one

`p` is a shortcut for the coin-flip walk. For anything else, pass a distribution as
`step_dist` and the walk adds up draws from it instead.

```python
RandomWalk(step_dist=Normal(0, 1)).sim(20).plot()
```

With `Normal(0, 1)` steps the variance after $n$ steps is still $n$, because each step still
has variance 1 -- the walk just moves smoothly instead of in jumps.

```python
round(RandomWalk(step_dist=Normal(0, 1))[100].sim(10000).var(), 1)
```

Steps do not have to be able to go down at all. With `Poisson(2)` steps the walk only ever
climbs, gaining 2 per step on average.

```python
RandomWalk(step_dist=Poisson(2)).sim(20).plot()
```
