## Conditional Random Variables

Simulating from a conditional distribution uses rejection sampling internally — draws that don't satisfy the condition are discarded. This naturally increases runtime, making the progress bar especially useful here.

```python
# Conditional RV — X | (X > 0) where X ~ Normal(0, 1)
# ~50% of draws are rejected, so each accepted draw takes ~2 raw draws on average.
# The combination of rejection sampling + slow draw triggers the progress bar.
import time

slow_normal = ProbabilitySpace(lambda: (time.sleep(0.001), np.random.normal())[1])
X = RV(slow_normal)
results = (X | (X > 0)).sim(5000)
print(f"E[X | X > 0] ≈ {results.mean():.3f}")  # should be ≈ 0.798
```

## BoxModel

Simulating coin flips from a box model. The progress bar appears here because each draw is artificially slowed — in practice, a BoxModel over a very large list or a complex draw function would have the same effect.

```python
# BoxModel.sim() — sampling with replacement from a list
# Wrapping with a slow ProbabilitySpace to trigger the progress bar
import time

slow_box = ProbabilitySpace(lambda: (time.sleep(0.001), np.random.choice(["H", "T"]))[1])
results = slow_box.sim(10000)
results.tabulate()
```
