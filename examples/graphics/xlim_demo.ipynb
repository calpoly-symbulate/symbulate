# How the plotting window chooses itself

`Distribution.plot()` has no window argument. The x-axis is framed from the distribution itself, in two steps:

- a **fixed bound is shown as-is**, and an unbounded side is cut where 0.1% of the probability is left outside;
- that window is then **zoomed in on** if the probability fills less than half of it.

So a small distribution keeps its true endpoints, and a spread-out one is framed on the part you can actually read. It works the same for `cdf=True`, since the window only picks the x-values.

```python
from symbulate import *
```

## A distribution that fills its window keeps every bound

All ten outcomes are on screen, including the rare 0 and 10.

```python
Binomial(10, 0.5).plot()
```

## A spread-out one zooms in by itself

The same distribution with a thousand trials puts everything in a band about a tenth of `(0, 1000)` wide. Drawn over the whole support it would be a spike in an empty axis, so the band is framed instead -- with nothing passed in.

```python
Binomial(1000, 0.5).plot()
```

## The same on a continuous distribution

`Beta(2, 200)` lives in the first few percent of `(0, 1)`.

```python
Beta(2, 200).plot()
```

## A fixed lower bound is given up only when nothing is near it

`Poisson(3)` starts at its true lower bound of 0. `Poisson(1000)` has the same bound, but nothing within 900 of it, so it is dropped.

```python
Poisson(3).plot()
```

```python
Poisson(1000).plot()
```

## Choosing the window yourself

Set it on the distribution before plotting. The curve is then computed across that range, which matters for a heavy tail: cropping afterwards would leave only a couple of grid points.

```python
X = HalfCauchy(scale=2)
X.xlim = (0, 10)
X.plot()
```

## Or move the axis after plotting

`xlim(low, high)` trims whatever is on screen -- handy when simulated values have stretched the axis out. The `;` just hides the pair of limits it returns.

```python
Normal(0, 1).plot(cdf=True)
xlim(-2, 2);
```
