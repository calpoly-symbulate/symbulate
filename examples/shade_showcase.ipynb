# Shading probability regions

Chain `.shade(...)` onto `.plot()` to fill a tail or interval under the curve. Bounds read as probability inequalities: `lt`/`le` set an upper bound (`X < lt` / `X <= le`), `gt`/`ge` set a lower bound (`X > gt` / `X >= ge`).

```python
from symbulate import *
```

## A left tail

```python
Normal(0, 1).plot().shade(lt=-1.96)
```

## A right tail

```python
Exponential(1).plot().shade(ge=2)
```

## An interval

```python
Normal(0, 1).plot().shade(gt=-1, lt=1)
```

## The whole visible curve

With no bounds, the entire curve is shaded.

```python
Gamma(2).plot().shade()
```

## Discrete: strict vs. inclusive

`lt` leaves the mass at the endpoint out; `le` includes it.

```python
Poisson(3).plot().shade(lt=3)   # 0, 1, 2
```

```python
Poisson(3).plot().shade(le=3)   # 0, 1, 2, 3
```

## A discrete interval

```python
Binomial(10, 0.5).plot().shade(gt=3, le=7)
```

## A discrete right tail

```python
Binomial(20, 0.5).plot().shade(ge=12)
```

## Shade several regions by chaining

```python
Normal(0, 1).plot().shade(lt=-1.96).shade(gt=1.96)
```

## Under a CDF

```python
Normal(0, 1).plot(cdf=True).shade(lt=0)
```

## Under a discrete CDF

```python
Poisson(3).plot(cdf=True).shade(le=2)
```

## Shading follows the plot's window

The plot frames the high-probability region by itself, and an open tail shades to that visible edge.

```python
Binomial(100, 0.5).plot().shade(le=50)
```
