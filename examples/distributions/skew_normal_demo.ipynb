# Skew-Normal Distribution

`SkewNormal(loc, scale, shape)` is a normal distribution that has been **tilted** to one side. 

```python
from symbulate import *
```

## Draw a value from a skew-normal distribution

```python
SkewNormal(loc = 5, scale = 2, shape = 4).draw()
```

## Theoretical mean, variance, and standard deviation


```python
X = SkewNormal(loc = 5, scale = 2, shape = 4)
X.mean(), X.var(), X.sd()
```

## Assign and simulate a random variable

```python
X = RV(SkewNormal(loc = 0, scale = 1, shape = 4))
X.sim(10000)
```

## Plot the pdf

```python
SkewNormal(loc = 5, scale = 2, shape = 4).plot()
```

## Plot the CDF

```python
SkewNormal(loc = 5, scale = 2, shape = 4).plot(cdf = True)
```

## A simulation matches the theoretical pdf

Simulate the random variable and overlay the true density on the histogram.

```python
RV(SkewNormal(loc = 0, scale = 1, shape = 4)).sim(10000).plot()
SkewNormal(loc = 0, scale = 1, shape = 4).plot()
```

## Both limiting cases

At `shape = 0` the tilt is gone and the distribution is exactly `Normal(loc, scale)`. As `shape` 
grows the left half is squeezed away and what is left is `HalfNormal`.

```python
SkewNormal(loc = 0, scale = 1, shape = 0).plot()
Normal(mean = 0, sd = 1).plot()
```

```python
SkewNormal(loc = 0, scale = 1, shape = 100).plot()
HalfNormal(scale = 1).plot()
```
