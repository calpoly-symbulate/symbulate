# Logistic Distribution

`Logistic(loc, scale)` is a continuous distribution that underlies logistic regression.

```python
from symbulate import *
```

## Draw a value from a Logistic distribution

```python
Logistic(loc=0, scale=1).draw()
```

## Theoretical mean, variance, and standard deviation

The mean equals `loc`; the standard deviation is `scale * pi / sqrt(3)`.

```python
Logistic(loc=0, scale=1).mean(), Logistic(loc=0, scale=1).var(), Logistic(loc=0, scale=1).sd()
```

## Assign and simulate a random variable

```python
L = RV(Logistic(loc=0, scale=1))
L.sim(10000)
```

## Plot the pdf

Centered at `loc`.

```python
Logistic(loc=0, scale=1).plot()
```

## Plot the CDF

This is the sigmoid function used in logistic regression.

```python
Logistic(loc=0, scale=1).plot(cdf=True)
```

## Scale parameter changes the spread

```python
for s in [0.5, 1, 2, 4]:
    Logistic(loc=0, scale=s).plot()
```

## Relationship: the log-odds of a Uniform

The logistic CDF is the sigmoid used in logistic regression, so its inverse -- the log-odds `log(U / (1 - U))` of a `Uniform(0, 1)` probability -- is `Logistic(0, 1)`.

```python
U = RV(Uniform(a=0, b=1))
(log(U / (1 - U))).sim(10000).plot()
Logistic(loc=0, scale=1).plot()
```
