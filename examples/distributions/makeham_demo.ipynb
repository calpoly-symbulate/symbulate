# Makeham (Gompertz-Makeham) Distribution

`Makeham(shape, makeham, scale)` is a continuous distribution used in actuarial science to model mortality. It is the Gompertz law plus a constant, age-independent hazard term (the "accident" term). Setting `makeham=0` recovers the plain Gompertz distribution.

```python
from symbulate import *
```

## Draw a value from a Makeham distribution

```python
Makeham(shape=1.5, makeham=0.3, scale=2).draw()
```

## Theoretical mean, variance, and standard deviation

```python
Makeham(shape=1.5, makeham=0.3, scale=2).mean(), Makeham(shape=1.5, makeham=0.3, scale=2).var(), Makeham(shape=1.5, makeham=0.3, scale=2).sd()
```

## Assign and simulate a random variable

```python
M1 = RV(Makeham(shape=1.5, makeham=0.3, scale=2))
M1.sim(10000)
```

## Plot the pdf

```python
Makeham(shape=1.5, makeham=0.3, scale=2).plot()
```

## Plot the CDF

```python
Makeham(shape=1.5, makeham=0.3, scale=2).plot(cdf=True)
```

## The makeham term adds a constant floor to mortality

With `makeham=0` this is exactly a Gompertz. Larger values raise the age-independent risk, shifting mass toward smaller values.

```python
for a in [0, 0.3, 1, 2]:
    Makeham(shape=1.5, makeham=a, scale=1).plot()
```

## Relationship: `makeham = 0` is exactly a Gompertz

With no accident term the two distributions coincide.

```python
Makeham(shape=1.5, makeham=0, scale=2).plot()
Gompertz(shape=1.5, scale=2).plot()
```
