# Noncentral t, chi-square, and F

The `StudentT`, `ChiSquare`, and `F` distributions each take an optional `noncentrality` parameter. With `noncentrality = 0` (the default) each reduces to its ordinary central distribution. A nonzero value is the distribution the test statistic follows when the null hypothesis is false -- the basis of power and effect-size analysis.

```python
from symbulate import *
```

## Noncentral t

The noncentrality shifts the t-distribution off center and makes it asymmetric.

```python
StudentT(df=10).plot()
StudentT(df=10, noncentrality=3).plot()
```

```python
StudentT(df=10).mean(), StudentT(df=10, noncentrality=3).mean()
```

## Noncentral chi-square

The noncentrality shifts the chi-square to the right.

```python
ChiSquare(df=4).plot()
ChiSquare(df=4, noncentrality=6).plot()
```

```python
# mean = df + noncentrality
ChiSquare(df=4).mean(), ChiSquare(df=4, noncentrality=6).mean()
```

## Noncentral F

This is the distribution the ANOVA F-statistic follows when the group means differ.

```python
F(dfN=5, dfD=20).plot()
F(dfN=5, dfD=20, noncentrality=10).plot()
```

```python
F(dfN=5, dfD=20).mean(), F(dfN=5, dfD=20, noncentrality=10).mean()
```

## Everything else still works

A noncentral distribution can still draw a value, define a random variable, simulate, and plot.

```python
ChiSquare(df=4, noncentrality=6).draw()
```

```python
RV(ChiSquare(df=4, noncentrality=6)).sim(10000).plot()
ChiSquare(df=4, noncentrality=6).plot()
```

```python
ChiSquare(df=4, noncentrality=6).plot(cdf=True)
```

## Relationship: the square of a t is an F

If `T` has a `StudentT(df)` distribution, then `T ** 2` has an `F(1, df)` distribution.

```python
T = RV(StudentT(df=10))
(T ** 2).sim(10000).plot()
F(dfN=1, dfD=10).plot()
```
