# Log-Uniform (Reciprocal) Distribution

`LogUniform(a, b)` is a continuous distribution on `[a, b]` that is uniform on a **logarithmic** scale: `log(X)` is uniform between `log(a)` and `log(b)`. Its density is proportional to `1 / x`.

```python
from symbulate import *
```

## Draw a value from a log-uniform distribution

```python
LogUniform(a=1, b=100).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `(b - a) / log(b / a)`; the median is the geometric mean `sqrt(a * b)` (here `10`).

```python
LogUniform(a=1, b=100).mean(), LogUniform(a=1, b=100).var(), LogUniform(a=1, b=100).sd()
```

## Assign and simulate a random variable

```python
L = RV(LogUniform(a=1, b=100))
L.sim(10000)
```

## Plot the pdf

The density falls off like `1 / x`, so smaller values are more likely.

```python
LogUniform(a=1, b=100).plot()
```

## Plot the CDF

```python
LogUniform(a=1, b=100).plot(cdf=True)
```

## Relationship: the log is Uniform

If `X` is `LogUniform(a, b)`, then `log(X)` is `Uniform(log(a), log(b))` -- the defining property.

```python
L = RV(LogUniform(a=1, b=100))
log(L).sim(10000).plot()
Uniform(a=log(1), b=log(100)).plot()
```

## Relationship: exponentiate a Uniform

Going the other way, `exp` of a `Uniform(log(a), log(b))` gives a `LogUniform(a, b)`.

```python
U = RV(Uniform(a=log(1), b=log(100)))
exp(U).sim(10000).plot(bins=100)
LogUniform(a=1, b=100).plot()
```

## The `Reciprocal` alias

`Reciprocal(a, b)` is the same distribution under its other common name, so the curves coincide.

```python
Reciprocal(a=1, b=100).plot()
LogUniform(a=1, b=100).plot()
```
