# Burr (Burr Type XII) Distribution

`Burr(shape1, shape2, scale)` is a heavy-tailed continuous distribution on `[0, inf)`. Often used in actuarial science and economics.

```python
from symbulate import *
```

## Draw a value from a Burr distribution

```python
Burr(shape1=3, shape2=2, scale=1).draw()
```

## Theoretical mean, variance, and standard deviation

The `k`-th moment is finite only when `shape1 * shape2 > k`, so a heavy-tailed Burr can have an infinite mean or variance.

```python
Burr(shape1=3, shape2=2, scale=1).mean(), Burr(shape1=3, shape2=2, scale=1).var(), Burr(shape1=3, shape2=2, scale=1).sd()
```

## Assign and simulate a random variable

```python
B = RV(Burr(shape1=3, shape2=2, scale=1))
B.sim(10000)
```

## Plot the pdf

```python
Burr(shape1=3, shape2=2, scale=1).plot()
```

## Plot the CDF

```python
Burr(shape1=3, shape2=2, scale=1).plot(cdf=True)
```

## Shape parameters change the curve

`shape1` shapes the body near the origin; `shape2` controls the tail.

```python
for shape1, shape2 in [(1, 2), (2, 1), (3, 2), (4, 4)]:
    Burr(shape1=shape1, shape2=shape2, scale=1).plot()
```

## Relationship: `shape1 = 1` is a Pareto

With `shape1 = 1` the Burr is a Pareto (Type II / Lomax) on `[0, inf)`. Shifting those values up by the scale gives a standard Pareto (Type I). The Burr generalizes the Pareto.

```python
P = RV(Burr(shape1=1, shape2=2, scale=1))
(P + 1).sim(10000).plot(bins=100)
Pareto(shape=2, scale=1).plot()
```
