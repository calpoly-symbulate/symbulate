# Scaled Inverse Chi-Square Distribution

`ScaledInverseChiSquare(df, scale)` is a continuous distribution on `(0, inf)`, it is an `InverseGamma` in a different parameterization -- `ScaledInverseChiSquare(df, scale) == InverseGamma(shape=df/2, scale=df*scale/2)`.

```python
from symbulate import *
```

## Draw a value from a scaled inverse chi-square distribution

```python
ScaledInverseChiSquare(df=6, scale=2).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `df * scale / (df - 2)` (finite only when `df > 2`); the variance is finite only when `df > 4`.

```python
ScaledInverseChiSquare(df=6, scale=2).mean(), ScaledInverseChiSquare(df=6, scale=2).var(), ScaledInverseChiSquare(df=6, scale=2).sd()
```

## Assign and simulate a random variable

```python
V = RV(ScaledInverseChiSquare(df=6, scale=2))
V.sim(10000)
```

## Plot the pdf

```python
ScaledInverseChiSquare(df=6, scale=2).plot()
```

## Plot the CDF

```python
ScaledInverseChiSquare(df=6, scale=2).plot(cdf=True)
```

## Degrees of freedom change the curve

A smaller `df` gives a heavier right tail and more spread; a larger `df` concentrates the mass.

```python
for df in [6, 10, 20, 40]:
    ScaledInverseChiSquare(df=df, scale=1).plot()
```

## Relationship: it is an Inverse Gamma

`ScaledInverseChiSquare(df, scale)` is exactly `InverseGamma(shape=df/2, scale=df*scale/2)`, so the two densities overlap.

```python
ScaledInverseChiSquare(df=6, scale=2).plot()
InverseGamma(shape=3, scale=6).plot()
```

## Relationship: `df * scale / X` is a chi-square

The name comes from this: if `X` has a scaled inverse chi-square distribution, then `df * scale / X` has a `ChiSquare(df)` distribution.

```python
X = RV(ScaledInverseChiSquare(df=6, scale=2))
(6 * 2 / X).sim(10000).plot()
ChiSquare(df=6).plot()
```
