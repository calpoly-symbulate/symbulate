# Beta-Binomial Distribution

`BetaBinomial(n, shape1, shape2)` counts the number of successes in `n` trials when the success probability is not fixed but is itself random.

```python
from symbulate import *
```

## Draw a value from a beta-binomial distribution

```python
BetaBinomial(n=10, shape1=2, shape2=3).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `n * shape1 / (shape1 + shape2)`.

```python
BetaBinomial(n=10, shape1=2, shape2=3).mean(), BetaBinomial(n=10, shape1=2, shape2=3).var(), BetaBinomial(n=10, shape1=2, shape2=3).sd()
```

## Assign and simulate a random variable

```python
B = RV(BetaBinomial(n=10, shape1=2, shape2=3))
B.sim(10000)
```

## Plot the pmf

```python
BetaBinomial(n=10, shape1=2, shape2=3).plot()
```

## Plot the CDF

```python
BetaBinomial(n=10, shape1=2, shape2=3).plot(cdf=True)
```

## Relationship: `shape1 = shape2 = 1` is a discrete uniform

```python
BetaBinomial(n=10, shape1=1, shape2=1).plot()
DiscreteUniform(a=0, b=10).plot()
```

## Relationship: large `shape1` and `shape2` approach a Binomial

```python
BetaBinomial(n=10, shape1=300, shape2=700).plot()
Binomial(n=10, p=0.3).plot()
```
