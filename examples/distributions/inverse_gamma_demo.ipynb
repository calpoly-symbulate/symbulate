# Inverse Gamma Distribution

`InverseGamma(shape, scale)` is a continuous distribution on `(0, inf)`. If `X` has a gamma distribution, then `1 / X` has an inverse gamma distribution.

```python
from symbulate import *
```

## Draw a value from an inverse gamma distribution

```python
InverseGamma(shape=3, scale=2).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `scale / (shape - 1)` and is finite only when `shape > 1`; the variance is finite only when `shape > 2`.

```python
InverseGamma(shape=3, scale=2).mean(), InverseGamma(shape=3, scale=2).var(), InverseGamma(shape=3, scale=2).sd()
```

## Assign and simulate a random variable

```python
V = RV(InverseGamma(shape=3, scale=2))
V.sim(10000)
```

## Plot the pdf

```python
InverseGamma(shape=3, scale=2).plot()
```

## Plot the CDF

```python
InverseGamma(shape=3, scale=2).plot(cdf=True)
```

## Shape parameter changes the curve

A smaller `shape` gives a heavier right tail and more spread; a larger `shape` concentrates near the origin.

```python
for shape in [3, 5, 8, 12]:
    InverseGamma(shape=shape, scale=1).plot()
```

## Relationship: the reciprocal of a Gamma

If `X` is `Gamma(shape, rate=scale)`, then `1 / X` is `InverseGamma(shape, scale)` --> The inverse gamma's scale equals the gamma's rate.

```python
G = RV(Gamma(shape=3, rate=2))
(1 / G).sim(10000).plot(bins=100)
InverseGamma(shape=3, scale=2).plot()
```
