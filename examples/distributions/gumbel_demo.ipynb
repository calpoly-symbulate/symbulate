# Gumbel (Extreme Value Type I) Distribution

`Gumbel(loc, scale)` is a continuous distribution on all real numbers for the maximum of many observations. It is exactly the `shape = 0` case of the `GEV` distribution.

```python
from symbulate import *
```

## Draw a value from a Gumbel distribution

```python
Gumbel(loc=0, scale=1).draw()
```

## Theoretical mean, variance, and standard deviation

The mean is `loc + euler_gamma * scale` (the Euler-Mascheroni constant is about 0.5772).

```python
Gumbel(loc=0, scale=1).mean(), Gumbel(loc=0, scale=1).var(), Gumbel(loc=0, scale=1).sd()
```

## Assign and simulate a random variable

```python
G = RV(Gumbel(loc=0, scale=1))
G.sim(10000)
```

## Plot the pdf

```python
Gumbel(loc=0, scale=1).plot()
```

## Plot the CDF

```python
Gumbel(loc=0, scale=1).plot(cdf=True)
```

## Scale parameter changes the spread

```python
for s in [0.5, 1, 2, 4]:
    Gumbel(loc=0, scale=s).plot()
```

## Relationship: Gumbel is the `shape = 0` GEV

`Gumbel(loc, scale)` is exactly `GEV(loc, scale, shape=0)`.

```python
Gumbel(loc=0, scale=1).plot()
GEV(loc=0, scale=1, shape=0).plot()
```

## Relationship: negative log of an Exponential

If `E` is `Exponential(rate=1)`, then `-log(E)` has a standard `Gumbel(0, 1)` distribution.

```python
E = RV(Exponential(rate=1))
(-log(E)).sim(10000).plot()
Gumbel(loc=0, scale=1).plot()
```

## Relationship: the difference of two Gumbels is Logistic

If `G1` and `G2` are independent `Gumbel(0, 1)` variables, then `G1 - G2` is `Logistic(0, 1)`.

```python
G1, G2 = RV(Gumbel(0, 1) * Gumbel(0, 1))
(G1 - G2).sim(10000).plot()
Logistic(loc=0, scale=1).plot()
```
