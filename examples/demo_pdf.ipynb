# Discrete PMF: dashed connecting line

`Binomial(n, 0.4)` shown two ways at three sizes, so the dashed connector can be judged against a small vs. large number of possible values:

- **Without overlay** — the true PMF alone: `Binomial(n, 0.4).plot()`. A filled dot marks each pmf value; the dashed line connects the dots (dashed so it can't be read as a continuous curve).
- **With overlay** — a simulation with the true PMF drawn on top. Two `.plot()` calls in one cell share the axes, so the true PMF lands over the simulated values.

`.sim()` needs an explicit count, so the overlay cells use `.sim(10000)`.

```python
from symbulate import *
```

## Small number of values — `Binomial(10, 0.4)` (11 possible values)

**Without overlay** — true PMF only

```python
Binomial(10, 0.4).plot()
```

**With overlay** — simulation with the true PMF on top

```python
RV(Binomial(10, 0.4)).sim(10000).plot()
Binomial(10, 0.4).plot()
```

## Medium number of values — `Binomial(100, 0.4)` (101 possible values)

**Without overlay** — true PMF only

```python
Binomial(100, 0.4).plot()
```

**With overlay** — simulation with the true PMF on top

```python
RV(Binomial(100, 0.4)).sim(10000).plot()
Binomial(100, 0.4).plot()
```

**Without overlay** — true PMF only

```python
Binomial(1000, 0.4).plot()
```
